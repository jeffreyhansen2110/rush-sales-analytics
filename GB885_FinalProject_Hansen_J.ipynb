```python
# looking at UNITS_SOLD for PRODUCT_ID = 20 in year 2021
# RESULT median = 232, mean = 355, high variability --> use median
df_sales[
    (df_sales['PRODUCT_ID'] == 20) &
    (df_sales['YEAR'] == 2021) &
    (df_sales['UNITS_SOLD'] != '***')
]['UNITS_SOLD'].astype(int).describe()
```

####Merging data

```python
# merge products with sales
df_merged = df_sales.merge(
    df_products,
    on='PRODUCT_ID',
    how='left'
)
```

```python
# verifying result
df_merged.shape
```

```python
# checking that every sales row is matched to a product name
df_merged['PRODUCT_NAME'].isna().sum()
```

```python
# adding in the cleaned retailer table
df_merged = df_merged.merge(
    df_retailer_clean,
    on='RETAILER_ID',
    how='left'
)
```

```python
# validate result
df_merged.shape
```

```python
# checking that every merged row has a retailer name
df_merged['RETAILER'].isna().sum()
```

####**SUMMARY of INSPECTION and CLEANING**
The three source tables were inspected, cleaned, and merged into a single analysis-ready dataframe.

Key cleaning steps included:
- Replacing two invalid UNITS_SOLD values using the median for comparable records
- Filling two missing PRICE_PER_UNIT values using nearby comparable transactions
- Converting INVOICE_DATE to datetime
- Correcting the SALES_METHOD typo "Ootlet" to "Outlet"
- Correcting one invalid RETAILER_ID
- Addressing duplicated retailer IDs to prevent duplication during the merge
- Preserving ambiguous Walmart/West Gear retailer IDs as an explicit ambiguous category

The final merged dataframe contains 9,648 rows, with no unmatched products or retailers.

##**Answering Business Questions**

```python
# total sales by product for 2021
sales_by_product_2021 = (
    df_merged[df_merged['YEAR'] == 2021]
    .groupby('PRODUCT_NAME')['TOTAL_SALES']
    .sum()
    .sort_values(ascending=False)
)

sales_by_product_2021
```

### VP Question 1: Highest-Selling Product Category in 2021

Men's Street Footwear generated the highest sales in 2021, with total sales of $74,667,440.

```python
# VP Question 2
womens_sales_by_state_2021 = (
    df_merged[
        (df_merged['YEAR'] == 2021) &
        (df_merged['PRODUCT_NAME'].str.contains("Women's"))
    ]
    .groupby('STATE')['TOTAL_SALES']
    .sum()
    .sort_values(ascending=False)
)

womens_sales_by_state_2021
```

### VP Question 2: Highest State Sales for Women's Products in 2021

Maine generated the highest sales of women's products in 2021, with total sales of $2,176,301.

```python
# VP Question 3
mens_sales_by_state_2021 = (
    df_merged[
        (df_merged['YEAR'] == 2021) &
        (df_merged['PRODUCT_NAME'].str.contains("Men's"))
    ]
    .groupby('STATE')['TOTAL_SALES']
    .sum()
    .sort_values(ascending=False)
)

mens_sales_by_state_2021
```

### VP Question 3: Highest State Sales for Men's Products in 2021

New Hampshire generated the highest sales of men's products in 2021, with total sales of $5,420,808.

```python
# VP Question 4
retailer_units = (
    df_merged[df_merged['YEAR'].isin([2020, 2021])]
    .groupby(['YEAR', 'RETAILER'])['UNITS_SOLD']
    .sum()
    .reset_index()
)

top_retailer_by_year = retailer_units.loc[
    retailer_units.groupby('YEAR')['UNITS_SOLD'].idxmax()
]

top_retailer_by_year
```

### VP Question 4: Retailer Purchasing the Most Units

- In 2020, Amazon purchased the most units, with 317,930 units.
- In 2021, Foot Locker purchased the most units, with 1,097,410 units.

```python
# overall sales growth 2020 to 2022
sales_by_year = (
    df_merged
    .groupby('YEAR')['TOTAL_SALES']
    .sum()
    .sort_index()
)

sales_by_year
```

```python
# growth rate calculaton
sales_growth_pct = (
    (sales_by_year.loc[2021] - sales_by_year.loc[2020])
    / sales_by_year.loc[2020]
) * 100

sales_growth_pct
```
